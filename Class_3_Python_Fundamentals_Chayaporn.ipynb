# Part 1: ตัวแปรและการเก็บข้อมูล

## 1.1 ตัวแปรคืออะไร?

**ตัวแปร (Variables)** คือ "กล่องเก็บข้อมูล" ในโปรแกรมของเรา เราใช้ตัวแปรในการเก็บค่าต่างๆ เพื่อนำไปใช้งานในโปรแกรม

### การสร้างตัวแปร

ใน Python การสร้างตัวแปรทำได้ง่ายมาก ไม่ต้องประกาศชนิดข้อมูลล่วงหน้า Python จะกำหนดชนิดข้อมูลให้อัตโนมัติตามค่าที่เราใส่

**รูปแบบ:**
```python
ชื่อตัวแปร = ค่าที่ต้องการเก็บ
```

````python
# ตัวอย่างการสร้างตัวแปร
x = 5                    # ตัวแปร x เก็บค่าตัวเลข 5
name = "Aichayaaaa"      # ตัวแปร name เก็บข้อความ
price = 99.99            # ตัวแปร price เก็บทศนิยม

# แสดงค่าตัวแปร
print("x =", x)
print("name =", name)
print("price =", price)
````

# Part 2: ชนิดข้อมูลพื้นฐาน

Python มีชนิดข้อมูลหลักๆ ที่ควรรู้ ได้แก่:

1. **Numbers** (ตัวเลข): int, float, complex
2. **Strings** (ข้อความ): str
3. **Booleans** (ค่าความจริง): bool

## 2.1 ตัวเลข (Numbers)

### 2.1.1 จำนวนเต็ม (Integer: int)
เก็บค่าตัวเลขที่เป็นจำนวนเต็มทั้งบวกและลบ ไม่มีทศนิยม

````python
# ตัวอย่างจำนวนเต็ม
student_count = 45
temperature = -5
year = 2024

print("จำนวนนักเรียน:", student_count)
print("ชนิดข้อมูล:", type(student_count))

# การคำนวณด้วยจำนวนเต็ม
a = 10
b = 3
print(f"{a} + {b} = {a + b}")
print(f"{a} - {b} = {a - b}")
print(f"{a} * {b} = {a * b}")
print(f"{a} / {b} = {a / b}")      # หารได้ float
print(f"{a} // {b} = {a // b}")    # หารปัดเศษ (floor division)
print(f"{a} % {b} = {a % b}")      # เศษจากการหาร (modulo)
print(f"{a} ** {b} = {a ** b}")    # ยกกำลัง
````

### 2.1.2 ทศนิยม (Float)
เก็บค่าตัวเลขที่มีจุดทศนิยม

````python
# ตัวอย่างทศนิยม
pi = 3.14159
price = 49.99
temperature = 36.5

print("ค่า pi =", pi)
print("ชนิดข้อมูล:", type(pi))

# การคำนวณพื้นที่วงกลม
radius = 5.0
area = pi * (radius ** 2)
print(f"รศมี {radius} cm มีพื้นที่ {area:.10f} ตารางเซนติเมตร")
````

### 2.1.3 จำนวนเชิงซ้อน (Complex)
เก็บค่าจำนวนที่มีทั้งส่วนจริงและส่วนจินตภาพ (ใช้ในทางคณิตศาสตร์ขั้นสูง)

````python
# ตัวอย่างจำนวนเชิงซ้อน
z = 3 + 4j
print("จำนวนเชิงซ้อน:", z)
print("ชนิดข้อมูล:", type(z))
print("ส่วนจริง:", z.real)
print("ส่วนจินตภาพ:", z.imag)
````

## 2.2 สตริง (Strings)

**สตริง** คือลำดับของอักขระ (characters) ใช้เก็บข้อความ

### การสร้างสตริง
ใช้เครื่องหมายคำพูดเดี่ยว `'...'` หรือคำพูดคู่ `"..."`

````python
# ตัวอย่างการสร้างสตริง
name = "สมชาย ใจดี"
message = 'สวัสดีครับ'
multi_line = """นี่คือข้อความ
ที่มีหลายบรรทัด
สามารถขึ้นบรรทัดใหม่ได้"""

print(name)
print(message)
print(multi_line)
print("ชนิดข้อมูล:", type(name))
````

### การเชื่อมต่อสตริง (String Concatenation)

````python
# การเชื่อมสตริงด้วย +
first_name = "สมชาย"
last_name = "รักเรียน"
full_name = first_name + " " + last_name
print("ชื่อเต็ม:", full_name)

# การทำซ้ำสตริงด้วย *
separator = "-" * 30
print(separator)
print("หัวข้อข่าว")
print(separator)
````

## 2.3 บูลีน (Booleans)

**Boolean** เป็นชนิดข้อมูลที่มีค่าได้เพียง 2 ค่า คือ `True` (จริง) หรือ `False` (เท็จ)
ใช้ในการตรวจสอบเงื่อนไขและควบคุมการทำงานของโปรแกรม

````python
# ตัวอย่าง Boolean
is_student = True
is_graduated = False

print("เป็นนักเรียน:", is_student)
print("จบการศึกษา:", is_graduated)
print("ชนิดข้อมูล:", type(is_student))

# การเปรียบเทียบให้ผลเป็น Boolean
age = 20
print("อายุมากกว่า 18:", age > 18)
print("อายุเท่ากับ 20:", age == 20)
print("อายุน้อยกว่า 18:", age < 18)
````

## 2.4 การตรวจสอบและแปลงชนิดข้อมูล

### ตรวจสอบชนิดข้อมูลด้วย `type()`

````python
# ตรวจสอบชนิดข้อมูล
x = 10
y = 3.14
z = "Hello"
w = True

print(f"x = {x}, ชนิด: {type(x)}")
print(f"y = {y}, ชนิด: {type(y)}")
print(f"z = {z}, ชนิด: {type(z)}")
print(f"w = {w}, ชนิด: {type(w)}")
````

### การแปลงชนิดข้อมูล (Type Casting)

````python
# แปลง int เป็น float
num_int = 10
num_float = float(num_int)
print(f"{num_int} (int) → {num_float} (float)")

# แปลง float เป็น int (ตัดทศนิยม)
pi = 3.14159
pi_int = int(pi)
print(f"{pi} (float) → {pi_int} (int)")

# แปลงเป็น string
age = 20
age_str = str(age)
message = "อายุของฉันคือ " + age_str + " ปี"
print(message)

# แปลง string เป็นตัวเลข
price_str = "99.99"
price = float(price_str)
print(f"ราคา: {price * 2} บาท")
````

# Part 3: การทำงานกับสตริง (String Operations)

สตริงเป็นชนิดข้อมูลที่ใช้บ่อยมากในการเขียนโปรแกรม Python มี methods มากมายสำหรับจัดการสตริง

## 3.1 การเข้าถึงตัวอักษรในสตริง (Indexing)

สตริงใน Python เก็บอักขระเป็นลำดับ เราสามารถเข้าถึงตัวอักษรแต่ละตัวได้ด้วย index (เริ่มที่ 0)

````python
# การเข้าถึงตัวอักษร
text = "Python"
print("ตัวอักษรแรก:", text[0])      # P
print("ตัวอักษรที่ 2:", text[1])     # y
print("ตัวอักษรสุดท้าย:", text[-1])  # n (นับจากท้าย)
print("ตัวอักษรสุดท้ายที่ 2:", text[-2])  # o

# การตัดสตริง (Slicing)
print("ตัวที่ 0-2:", text[0:3])      # Pyt
print("ตัวที่ 2 เป็นต้นไป:", text[2:])  # thon
print("ถึงตัวที่ 4:", text[:4])      # Pyth
````

## 3.2 String Methods สำคัญ

Python มี methods มากมายสำหรับจัดการสตริง

### 3.2.1 การเปลี่ยนตัวพิมพ์

````python
text = "Hello World"

print("ตัวพิมพ์ใหญ่ทั้งหมด:", text.upper())       # HELLO WORLD
print("ตัวพิมพ์เล็กทั้งหมด:", text.lower())       # hello world
print("ตัวแรกของแต่ละคำใหญ่:", text.title())     # Hello World
print("สลับตัวพิมพ์:", text.swapcase())           # hELLO wORLD
print("ตัวแรกพิมพ์ใหญ่:", text.capitalize())     # Hello world
````

### 3.2.2 การตัดช่องว่าง

````python
text = "   Hello World   "

print("ตัดช่องว่างทั้งสองข้าง:", text.strip())     # "Hello World"
print("ตัดช่องว่างซ้าย:", text.lstrip())           # "Hello World   "
print("ตัดช่องว่างขวา:", text.rstrip())           # "   Hello World"

# ตัวอย่างการใช้งานจริง: ทำความสะอาดข้อมูลที่ผู้ใช้ป้อน
user_input = "  Bangkok  "
clean_input = user_input.strip()
print(f"ข้อมูลที่ทำความสะอาด: '{clean_input}'")
````

### 3.2.3 การค้นหาและแทนที่

````python
text = "Python is easy. Python is powerful."

# ค้นหาตำแหน่ง
print("หา 'easy':", text.find("easy"))           # 10
print("หา 'difficult':", text.find("difficult")) # -1 (ไม่เจอ)

# ตรวจสอบว่ามีคำนี้หรือไม่
print("มีคำว่า 'Python' หรือไม่:", "Python" in text)  # True
print("มีคำว่า 'Java' หรือไม่:", "Java" in text)      # False

# แทนที่คำ
new_text = text.replace("Python", "JavaScript")
print("แทนที่:", new_text)
````

### 3.2.4 การแยกและรวมสตริง

````python
# split() - แยกสตริงเป็น list
sentence = "Python is easy to learn"
words = sentence.split()
print("แยกคำ:", words)

# แยกด้วยตัวคั่นที่กำหนด
csv_data = "Bangkok,Thailand,10539000"
city_info = csv_data.split(",")
print("ข้อมูลเมือง:", city_info)

# join() - รวม list เป็นสตริง
words = ["สวัสดี", "ครับ", "ยินดีต้อนรับ"]
greeting = " ".join(words)
print("ข้อความทักทาย:", greeting)

# รวมด้วยตัวคั่นอื่น
cities = ["Bangkok", "Chiang Mai", "Phuket"]
cities_str = ", ".join(cities)
print("เมืองต่างๆ:", cities_str)
````

### 3.2.5 การตรวจสอบลักษณะของสตริง

````python
# ตรวจสอบว่าเป็นตัวอักษร/ตัวเลข
text1 = "Python3"
text2 = "Python"
text3 = "12345"

print(f"'{text1}' เป็นตัวอักษรและตัวเลขหรือไม่:", text1.isalnum())  # True
print(f"'{text2}' เป็นตัวอักษรทั้งหมดหรือไม่:", text2.isalpha())    # True
print(f"'{text3}' เป็นตัวเลขทั้งหมดหรือไม่:", text3.isdigit())      # True

# ตรวจสอบตัวพิมพ์
print("'HELLO' เป็นตัวพิมพ์ใหญ่ทั้งหมดหรือไม่:", "HELLO".isupper())
print("'hello' เป็นตัวพิมพ์เล็กทั้งหมดหรือไม่:", "hello".islower())
````

## 3.3 การจัดรูปแบบสตริง (String Formatting)

### 3.3.1 f-strings (แนะนำ - Python 3.6+)

````python
# f-string พื้นฐาน
name = "สมชาย"
age = 20
height = 175.5

# แทรกค่าตัวแปร
message = f"ชื่อ: {name}, อายุ: {age} ปี, ส่วนสูง: {height} ซม."
print(message)

# คำนวณในสตริง
print(f"อีก 5 ปี {name} จะอายุ {age + 5} ปี")

# จัดรูปแบบทศนิยม
pi = 3.14159265359
print(f"ค่า pi = {pi:.2f}")          # 2 ตำแหน่งทศนิยม
print(f"ค่า pi = {pi:.4f}")          # 4 ตำแหน่งทศนิยม

# จัดรูปแบบตัวเลขมีคอมม่า
population = 10539000
print(f"ประชากร: {population:,} คน")
````

### 3.3.2 .format() method

````python
# ใช้ .format()
name = "สมหญิง"
gpa = 3.75

message = "นักศึกษา {} มี GPA {:.2f}".format(name, gpa)
print(message)

# ใช้ตำแหน่ง
text = "{0} + {1} = {2}".format(5, 3, 5+3)
print(text)

# ใช้ชื่อตัวแปร
text = "เมือง: {city}, ประเทศ: {country}".format(city="Bangkok", country="Thailand")
print(text)
````

# Part 4: โครงสร้างข้อมูล (Data Structures)

Python มีโครงสร้างข้อมูลหลัก 4 ชนิดที่ใช้เก็บข้อมูลหลายค่า:

1. **List** (ลิสต์) - เก็บข้อมูลหลายค่า แก้ไขได้
2. **Tuple** (ทูเพิล) - เก็บข้อมูลหลายค่า แก้ไขไม่ได้
3. **Set** (เซ็ต) - เก็บข้อมูลไม่ซ้ำ ไม่มีลำดับ
4. **Dictionary** (ดิกชันนารี) - เก็บข้อมูลแบบคู่ key-value

## 4.1 List (ลิสต์)

**List** เป็นโครงสร้างข้อมูลที่เก็บค่าหลายค่าในตัวแปรเดียว สามารถเปลี่ยนแปลงได้ (mutable) และมีลำดับ

### การสร้าง List

````python
# สร้าง list
fruits = ["apple", "banana", "orange"]
numbers = [1, 2, 3, 4, 5]
mixed = [1, "hello", 3.14, True]  # สามารถเก็บหลายชนิดได้
empty_list = []  # list ว่าง

print("ผลไม้:", fruits)
print("ตัวเลข:", numbers)
print("ข้อมูลผสม:", mixed)
print("ชนิดข้อมูล:", type(fruits))
````

### การเข้าถึงข้อมูลใน List

````python
fruits = ["apple", "banana", "orange", "mango", "grape"]

# เข้าถึงด้วย index (เริ่มที่ 0)
print("ตัวแรก:", fruits[0])      # apple
print("ตัวที่สอง:", fruits[1])    # banana
print("ตัวสุดท้าย:", fruits[-1])  # grape (นับจากท้าย)

# Slicing
print("3 ตัวแรก:", fruits[:3])   # ['apple', 'banana', 'orange']
print("ตัวที่ 2-4:", fruits[1:4]) # ['banana', 'orange', 'mango']
print("ทุกตัวที่สอง:", fruits[::2])  # ['apple', 'orange', 'grape']
````

### การแก้ไขข้อมูลใน List

````python
fruits = ["apple", "banana", "orange"]
print("ก่อนแก้ไข:", fruits)

# แก้ไขค่า
fruits[1] = "mango"
print("หลังแก้ไข:", fruits)

# เพิ่มข้อมูล
fruits.append("grape")        # เพิ่มท้ายสุด
print("หลัง append:", fruits)

fruits.insert(1, "kiwi")      # แทรกตำแหน่งที่ 1
print("หลัง insert:", fruits)

# ลบข้อมูล
fruits.remove("orange")       # ลบตามค่า
print("หลัง remove:", fruits)

deleted = fruits.pop()        # ลบตัวสุดท้าย และส่งค่ากลับ
print("ลบตัวสุดท้าย:", deleted)
print("หลัง pop:", fruits)
````

### List Methods ที่ใช้บ่อย

````python
numbers = [3, 1, 4, 1, 5, 9, 2, 6]

# ความยาวของ list
print("จำนวนสมาชิก:", len(numbers))

# หาค่าต่างๆ
print("ค่าสูงสุด:", max(numbers))
print("ค่าต่ำสุด:", min(numbers))
print("ผลรวม:", sum(numbers))

# นับจำนวนครั้งที่ปรากฏ
print("1 ปรากฏ:", numbers.count(1), "ครั้ง")

# หาตำแหน่ง
print("ตำแหน่งของ 5:", numbers.index(5))

# เรียงลำดับ
numbers.sort()  # เรียงจากน้อยไปมาก
print("เรียงแล้ว:", numbers)

numbers.sort(reverse=True)  # เรียงจากมากไปน้อย
print("เรียงย้อนกลับ:", numbers)

# กลับลำดับ
numbers.reverse()
print("กลับลำดับ:", numbers)
````

**Tuple** คล้าย List แต่**ไม่สามารถแก้ไขได้** (immutable) หลังจากสร้างแล้ว ใช้เก็บข้อมูลที่ไม่ต้องการให้เปลี่ยนแปลง

### การสร้าง Tuple

````python
# สร้าง tuple ด้วยวงเล็บ ()
point = (13.7563, 100.5018)  # พิกัดกรุงเทพฯ
rgb_color = (255, 0, 0)      # สีแดง
single_item = (42,)           # tuple ที่มีค่าเดียว (ต้องมีคอมม่า)

print("พิกัด:", point)
print("สี RGB:", rgb_color)
print("ชนิดข้อมูล:", type(point))
````

### การเข้าถึงข้อมูลใน Tuple

````python
point = (13.7563, 100.5018)

# เข้าถึงด้วย index (เหมือน list)
latitude = point[0]
longitude = point[1]
print(f"ละติจูด: {latitude}, ลองจิจูด: {longitude}")

# Unpacking (แกะค่า)
lat, lon = point
print(f"Unpacked - lat: {lat}, lon: {lon}")

# ไม่สามารถแก้ไขได้ (จะ Error)
# point[0] = 15.0  # TypeError: 'tuple' object does not support item assignment
````

## 4.3 Set (เซ็ต)

**Set** เป็นโครงสร้างข้อมูลที่เก็บค่าที่**ไม่ซ้ำกัน** และ**ไม่มีลำดับ**

### การสร้าง Set

````python
# สร้าง set ด้วยวงเล็บปีกกา {}
fruits = {"apple", "banana", "orange"}
numbers = {1, 2, 3, 4, 5}

# สร้างจาก list (จะลบค่าซ้ำอัตโนมัติ)
numbers_with_duplicates = [1, 2, 2, 3, 3, 3, 4, 5]
unique_numbers = set(numbers_with_duplicates)

print("ผลไม้:", fruits)
print("ตัวเลขเดิม:", numbers_with_duplicates)
print("ตัวเลขไม่ซ้ำ:", unique_numbers)
````

### การดำเนินการกับ Set

````python
regions = {"Asia", "Europe", "Africa"}

# เพิ่มสมาชิก
regions.add("America")
print("เพิ่ม America:", regions)

# พยายามเพิ่มซ้ำ (ไม่มีผล)
regions.add("Asia")
print("เพิ่ม Asia ซ้ำ:", regions)  # ไม่เปลี่ยนแปลง

# ลบสมาชิก
regions.remove("Africa")  # ลบ (ถ้าไม่มีจะ Error)
print("ลบ Africa:", regions)

regions.discard("Antarctica")  # ลบ (ถ้าไม่มีก็ไม่ Error)

# ตรวจสอบว่ามีสมาชิกหรือไม่
print("มี Asia หรือไม่:", "Asia" in regions)
print("จำนวนสมาชิก:", len(regions))
````

### Set Operations (การดำเนินการทางคณิตศาสตร์)

````python
# สองกลุ่มนักเรียน
math_students = {"สมชาย", "สมหญิง", "สมศรี", "สมพงษ์"}
science_students = {"สมหญิง", "สมศรี", "สมปอง", "สมหมาย"}

# Union (รวมกัน)
all_students = math_students | science_students
print("นักเรียนทั้งหมด:", all_students)

# Intersection (ตัดกัน - เรียนทั้งสองวิชา)
both_subjects = math_students & science_students
print("เรียนทั้งสองวิชา:", both_subjects)

# Difference (ต่างกัน - เรียนเฉพาะคณิตศาสตร์)
only_math = math_students - science_students
print("เรียนเฉพาะคณิตศาสตร์:", only_math)

# Symmetric Difference (เรียนเพียงวิชาเดียว)
one_subject_only = math_students ^ science_students
print("เรียนเพียงวิชาเดียว:", one_subject_only)
````

## 4.4 Dictionary (ดิกชันนารี)

**Dictionary** เก็บข้อมูลเป็นคู่ **key-value** โดย key ต้องไม่ซ้ำกัน ใช้ในการเก็บข้อมูลที่มีการเชื่อมโยง

### การสร้าง Dictionary

````python
# สร้าง dictionary ด้วย {key: value}
student = {
    "name": "สมชาย",
    "age": 20,
    "major": "Computer Science",
    "gpa": 3.75
}

print("ข้อมูลนักเรียน:", student)
print("ชนิดข้อมูล:", type(student))

# Dictionary ว่าง
empty_dict = {}
another_empty = dict()
````

### การเข้าถึงข้อมูลใน Dictionary

````python
student = {
    "name": "สมชาย",
    "age": 20,
    "major": "Geography"
}

# เข้าถึงด้วย key
print("ชื่อ:", student["name"])
print("อายุ:", student["age"])

# ใช้ .get() (ปลอดภัยกว่า ไม่ Error ถ้าไม่มี key)
print("สาขา:", student.get("major"))
print("GPA:", student.get("gpa", "ไม่มีข้อมูล"))  # ถ้าไม่มีให้ค่า default

# แก้ไขค่า
student["age"] = 21
print("อายุใหม่:", student["age"])

# เพิ่ม key-value ใหม่
student["gpa"] = 3.75
print("เพิ่ม GPA:", student)
````

### Dictionary Methods ที่ใช้บ่อย

````python
city = {
    "name": "Bangkok",
    "country": "Thailand",
    "population": 10_539_000,
    "area_km2": 1568.7
}

# ดู keys ทั้งหมด
print("Keys:", city.keys())

# ดู values ทั้งหมด
print("Values:", city.values())

# ดูคู่ key-value
print("Items:", city.items())

# วนลูปแสดงข้อมูล
print("\nข้อมูลเมือง:")
for key, value in city.items():
    print(f"  {key}: {value}")

# ลบ key
deleted_value = city.pop("area_km2")
print(f"\nลบพื้นที่: {deleted_value}")
print("Dictionary หลังลบ:", city)
````
