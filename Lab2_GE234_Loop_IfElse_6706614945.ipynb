## 🔹 ตัวอย่างที่ 1: การใช้ `for loop` เพื่อแสดงค่าพิกัด


```python

# พิกัดของสถานที่สำคัญในประเทศไทย
locations = {
    "Bangkok": (13.7563, 100.5018),
    "Chiang Mai": (18.7883, 98.9853),
    "Phuket": (7.8804, 98.3923),
    "Ubon Ratchathani": (15.2287, 104.8564)
}

# วนลูปแสดงค่าพิกัด
for city, coord in locations.items():
    print(f"{city}: Latitude {coord[0]}, Longitude {coord[1]}")
```


## 🔹 ตัวอย่างที่ 2: การใช้ `if-else` กรองข้อมูลพิกัด


```python

for city, coord in locations.items():
    if coord[0] > 16:
        print(f"{city} อยู่ทางภาคเหนือของประเทศไทย")
    else:
        print(f"{city} ไม่ใช่จังหวัดทางภาคเหนือ")
```


## 🔹 ตัวอย่างที่ 3: ใช้ `while loop` นับจำนวนจุดข้อมูล


```python

points = [(13.7563, 100.5018), (18.7883, 98.9853), (7.8804, 98.3923)]
count = 0

while count < len(points):
    print(f"Point {count + 1}: {points[count]}")
    count += 1
```


## 🔹 ตัวอย่างที่ 4: วิเคราะห์ข้อมูล Raster ด้วย `for loop`


```python

import numpy as np

# จำลองข้อมูล NDVI (Normalized Difference Vegetation Index)
ndvi = np.array([
    [0.2, 0.4, 0.5],
    [0.1, 0.3, 0.7],
    [0.6, 0.8, 0.2]
])

# วนลูปตรวจสอบค่าพืชพรรณ
for row in range(ndvi.shape[0]):
    for col in range(ndvi.shape[1]):
        if ndvi[row, col] > 0.3:
            print(f"พิกัด ({row}, {col}) มีพืชพรรณหนาแน่น (NDVI={ndvi[row, col]})")
```


# 📝 **กิจกรรมในแลป**

1. **แบบฝึกหัด 1**: เขียน `for loop` เพื่อแสดงชื่อจังหวัดและพิกัดจากลิสต์ของคุณเอง
2. **แบบฝึกหัด 2**: ใช้ `if-else` เพื่อแยกแยะพื้นที่ที่เป็นภูเขาและพื้นที่ราบจากลิสต์ของพิกัด
3. **แบบฝึกหัด 3**: ใช้ `while loop` เพื่อวนลูปและแสดงค่าละติจูดจากลิสต์จนกว่าจะพบค่าต่ำกว่า 10°
4. **แบบฝึกหัด 4**: ใช้ `for loop` วิเคราะห์ค่า NDVI และหาค่าพื้นที่ที่มีสภาพแห้งแล้ง (NDVI < 0.2)


แบบฝึกหัด 1

```python
provinces = [
    ("Bangkok", 13.7563, 100.5018),
    ("Chiang Mai", 18.7883, 98.9853),
    ("Phuket", 7.8804, 98.3923),
    ("Khon Kaen", 16.4419, 102.8350),
    ("Nakhon Ratchasima", 14.9799, 102.0977),
    ("Chonburi", 13.3611, 100.9847),
    ("Udon Thani", 17.4138, 102.7870),
    ("Surat Thani", 9.1401, 99.3331),
    ("Songkhla", 7.1898, 100.5951),
    ("Phitsanulok", 16.8211, 100.2659)
]

print("รายชื่อจังหวัดและพิกัด\n")

for i, (province, lat, lon) in enumerate(provinces, start=1):
    print(f"{i}. จังหวัด: {province}")
    print(f"   ละติจูด: {lat}")
    print(f"   ลองจิจูด: {lon}")
```

แบบฝึก 2

```python
heights = [120, 650, 300, 850, 45, 1000, 480]

print("วิเคราะห์ประเภทพื้นที่จากค่าความสูง\n")

for i, height in enumerate(heights, start=1):
    if height > 600:
        print(f"จุดที่ {i}: {height} เมตร -> พื้นที่ภูเขา")
    else:
        print(f"จุดที่ {i}: {height} เมตร -> พื้นที่ราบ")
```

แบบฝึก 3

```python
latitudes = [18.5, 15.2, 12.8, 9.7, 14.3, 8.9]

index = 0

print("แสดงค่าละติจูดจากลิสต์\n")

while index < len(latitudes):
    latitude = latitudes[index]

    if latitude < 10:
        print(f"พบค่าต่ำกว่า 10° ที่ค่า {latitude}°  หยุดการทำงาน")
        break

    print(f"ละติจูด: {latitude}°")
    index += 1
```

แบบฝึก 4

```python
import numpy as np

ndvi = np.array([
    [0.2, 0.4, 0.5],
    [0.1, 0.3, 0.7],
    [0.6, 0.8, 0.1]
])

for row in range(ndvi.shape[0]):
    for col in range(ndvi.shape[1]):
        if ndvi[row, col] < 0.2:
            print(f"พิกัด ({row}, {col}) แห้งแล้ง (NDVI = {ndvi[row, col]})")
```
